

# Student Performance Data Analysis




```python
import pandas as pd
```

```python
df = pd.read_csv('student-mat.csv', sep=';')

df.head()
```

## Dataset Overview

The dataset was loaded successfully using pandas.

```python
df.shape
```

```python
df.dtypes
```

## Dataset Information

The shape and datatypes of the dataset were checked to understand the structure of the data.

```python
df.isnull().sum()
```

```python
df = df.drop_duplicates()
```

### Data Cleaning

Missing values were checked and duplicate rows were removed.


## Data Analysis


```python
print("Average Final Grade:", float(round(df['G3'].mean(), 2)))
```

#### Average Final Grade (G3)

The average final grade of students was calculated using the mean() function.
The average final grade obtained was 10.42.

```python
print("Students scoring above 15:", (df['G3'] > 15).sum())
```

#### Students Scoring Above 15

The number of students who scored above 15 in the final grade column was calculated.
A total of 40 students scored above 15.

```python
correlation = df['studytime'].corr(df['G3'])

print("Correlation between study time and performance:", round(correlation, 2))
```

#### Correlation Between Study Time and Performance

Correlation analysis was performed to determine whether study time affects student performance.
The correlation value obtained was 0.1, indicating a weak positive relationship between study time and performance.

```python
gender_avg = df.groupby('sex')['G3'].mean().round(2)

print(gender_avg)
```

#### Gender-wise Performance Analysis

The average final grades of male and female students were compared using groupby() and mean() functions.

Female Students Average Grade: 9.97  
Male Students Average Grade: 10.91

Male students performed slightly better on average in this dataset.

## Data Visualization

Different graphs were created to visualize student performance patterns and relationships in the dataset.

```python
import matplotlib.pyplot as plt
```

```python
plt.hist(df['G3'])

plt.xlabel('Final Grade')
plt.ylabel('Number of Students')
plt.title('Distribution of Final Grades')

plt.show()
```

### Histogram of Final Grades

A histogram was created to visualize the distribution of final grades among students.
Most students scored between 8 and 15 marks, while fewer students scored very low or very high marks.

```python
plt.scatter(df['studytime'], df['G3'])

plt.xlabel('Study Time')
plt.ylabel('Final Grade')
plt.title('Study Time vs Final Grade')

plt.show()
```

### Scatter Plot of Study Time vs Final Grade

A scatter plot was created to visualize the relationship between study time and final grades.
The graph indicates a weak positive relationship between study time and student performance.

```python
gender_avg = df.groupby('sex')['G3'].mean()

gender_avg.plot(kind='bar')

plt.xlabel('Gender')
plt.ylabel('Average Final Grade')
plt.title('Average Grade by Gender')

plt.show()
```

### Gender-wise Performance Bar Chart

A bar chart was created to compare the average final grades of male and female students.
The graph shows that male students performed slightly better on average in this dataset.

## Conclusion

The student performance dataset was successfully analyzed using Python libraries such as pandas and matplotlib.

Data cleaning was performed by checking missing values and removing duplicate rows.

The average final grade of students was found to be 10.42, and 40 students scored above 15 marks.

Correlation analysis showed a weak positive relationship between study time and performance.

Visualizations helped in understanding grade distribution, study patterns, and gender-wise performance.
